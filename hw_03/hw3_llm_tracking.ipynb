```python
import os

os.environ["VLLM_HOST_IP"] = "127.0.0.1"

try:
    from vllm import LLM, SamplingParams
    
    print("vLLM импортирован успешно")
    
    llm = LLM(
        # модель поддерживает chat-template и следует инструкциям, а не просто продолжает текст
        model="Qwen/Qwen2.5-1.5B-Instruct",
        dtype="float16"
    )
    
    sampling_params = SamplingParams(temperature=0.8, max_tokens=20)
    outputs = llm.generate(["Hello, my name is"], sampling_params)
    
    print("Результат:", outputs[0].outputs[0].text)
    print("✅ vLLM работает!")
    
except Exception as e:
    print(f"❌ Ошибка: {e}")
    print("\nvLLM не поддерживается на этом железе или требует дополнительную настройку")
```

```python
import os
import mlflow
import pandas as pd
from openai import OpenAI
from mlflow.metrics.genai import make_genai_metric, EvaluationExample

os.environ["OPENAI_API_KEY"] = "dummy"
os.environ["OPENAI_API_BASE"] = "http://localhost:8000/v1"

client = OpenAI(base_url="http://localhost:8000/v1", api_key="dummy")

# Датафрейм с вопросами и эталонными ответами
eval_data = pd.DataFrame({
    "inputs": [
        "What is the capital of Germany?",
        "What is the capital of France?",
        "What is 2 + 2?",
        "Who wrote Hamlet?",
    ],
    "ground_truth": [
        "Berlin",
        "Paris",
        "4",
        "Shakespeare",
    ],
})

# Наполняем outputs ответами модели
def get_model_answer(question):
    response = client.chat.completions.create(
        model="Qwen/Qwen2.5-1.5B-Instruct",
        messages=[{"role": "user", "content": f"{question} Answer in one word."}],
        max_tokens=10,
        temperature=0.0,
    )
    return response.choices[0].message.content.strip()

eval_data["outputs"] = eval_data["inputs"].apply(get_model_answer)
print(eval_data[["inputs", "outputs", "ground_truth"]])
```

```python
mlflow.set_tracking_uri("http://localhost:5000")

# Метрика
correctness_metric = make_genai_metric(
    name="correctness",
    definition="Measures whether the answer matches the ground truth.",
    grading_prompt=(
        "Compare the answer with the ground truth. "
        "Score 5 if correct, 1 if wrong. Return only a number."
    ),
    grading_context_columns=["ground_truth"],
    examples=[
        EvaluationExample(
            input="What is the capital of Germany?",
            output="Berlin",
            score=5,
            justification="Matches ground truth.",
            grading_context={"ground_truth": "Berlin"},
        ),
        EvaluationExample(
            input="What is 2 + 2?",
            output="5",
            score=1,
            justification="Does not match ground truth.",
            grading_context={"ground_truth": "4"},
        ),
    ],
    model="openai:/Qwen/Qwen2.5-1.5B-Instruct",
    parameters={"temperature": 0.0, "max_tokens": 100},
    aggregations=["mean"],
    greater_is_better=True,
)

# Оценка
with mlflow.start_run(run_name="qwen2.5-1.5b-instruct-pipeline"):
    mlflow.log_param("model", "Qwen/Qwen2.5-1.5B-Instruct")
    mlflow.log_param("dataset_size", len(eval_data))

    results = mlflow.evaluate(
        data=eval_data,
        predictions="outputs",
        targets="ground_truth",
        extra_metrics=[correctness_metric],
        evaluator_config={"col_mapping": {"inputs": "inputs"}},
    )

    print(results.metrics)
    print(results.tables["eval_results_table"])
```
